For Loops and While Loops

```python
# for loop

fruits = ['apple', 'banana', 'orange']
for i in fruits :
    print('-', i)
```

```python
# for loop to count 10 to 20

for counting in range(11,21) :
    print(counting)
```

```python
name = 'art of war'
for naam in name: 
    print(naam, name)
```
